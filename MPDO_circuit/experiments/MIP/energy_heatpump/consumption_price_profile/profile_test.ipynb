```python
import numpy as np

def load_series(path):
    values = []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            _, val = line.split('\t')
            values.append(float(val))
    return np.array(values)

prices = load_series('prices.txt')
consumption = load_series('consumption.txt')
temperatures = load_series('temperatures.txt')

print(prices.shape, consumption.shape)  # (168,) (168,)
```

```python
import matplotlib.pyplot as plt

day = 6

plt.figure(figsize=(4,2))
plt.plot(prices[24 * (day-1): 24 * day])
plt.title('Prices')

plt.figure(figsize=(4,2))
plt.plot(consumption[24 * (day-1): 24 * day])
plt.title("Consumption")

plt.figure(figsize=(4,2))
plt.plot(temperatures[24 * (day-1): 24 * day])
plt.title('Temperatures')
```

```python
daily_consumption = []
for day in range(1,8):
    daily_consumption.append(np.sum(consumption[24 * (day-1): 24 * day]))

plt.figure(figsize=(4,2))
plt.plot(daily_consumption)
```
